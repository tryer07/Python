```python
#数据查看常见方法 --> head,tail,describe,info,shape,columns

import pandas as pd

df1 = pd.read_csv('data/sales.csv',usecols = ['订单号','产品类别','产品名称','销售数量','单价'])

print(df1.head(10))      #查看前10行数据
print(df1.tail(10))      #查看后10行数据
print(df1.describe())    #查看数据统计信息
df1.info()        #查看数据信息(列名，数据类型，非空值个数，内存使用情况)，该函数本身即打印到标准输出且返回None，无需再包print
print(df1.shape)         #查看数据形状
print(df1.columns)       #查看数据列名
```
