```python
import pandas as pd

df = pd.read_csv('data/sales.csv')

var1 = df

#数据过滤格式 --> df[df['过滤条件']]

#1.获取销售数量大于等于10的订单数据

var2 = df[df['销售数量']>=10]        # ✅ '销售' → '销售数量'

#2.获取产品类别为食品或者图书的订单数据

var3 = df[df['产品类别'].isin(['食品','图书'])]

#3.获取单价在100-200之间的订单数据

var4 = df[(df['单价']>=100) & (df['单价']<=200)]


#4.获取销售数量大于等于8且单价大于等于100的订单数据

var5 = df[(df['销售数量']>=8) & (df['单价']>=100)]


#5.获取产品类别为服装或者食品，支付方式为微信或者支付宝的订单数据

var6 = df[(df['产品类别'].isin(['服装','食品'])) & (df['支付方式'].isin(['微信支付','支付宝']))]   # ✅ '微信' → '微信支付'

print(var1)
print(var2)
print(var3)
print(var4)
print(var5)
print(var6)
```
