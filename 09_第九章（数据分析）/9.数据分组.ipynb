```python
import pandas as pd

#读取文件

df = pd.read_csv('data/sales.csv',nrows = 20)

#计算销售金额（销售数量 × 单价），供后续分组统计使用

df['销售金额'] = df['销售数量'] * df['单价']

#1.分组

#1.1根据产品类别分组，统计各个类别的订单数量

var1 = df.groupby('产品类别')['订单号'].count()

#1.2根据产品类别分组，统计各个类别的销售数量之和

var2 = df.groupby('产品类别')['销售数量'].sum()

#1.3根据产品类别分组，统计各个类别的销售金额之和

var3 = df.groupby('产品类别')['销售金额'].sum()

#1.4根据产品类别分组，统计各个类别的最低商品单价

var4 = df.groupby('产品类别')['单价'].min()

#1.5根据产品类别分组，统计各个类别的最高商品单价

var5 = df.groupby('产品类别')['单价'].max()

#1.6根据产品类别分组，统计各个类别的平均商品单价

var6 = df.groupby('产品类别')['单价'].mean()

#1.7根据产品类别分组，统计各个类别的最低商品单价、最高商品单价、平均商品单价

var7 = df.groupby('产品类别')['单价'].agg(['min', 'max', 'mean'])

#1.8根据产品类别分组，统计各个类别的商品的销售数量之和，销售金额之和，平均单价

var8 = df.groupby('产品类别').agg(
    销售数量之和=('销售数量', 'sum'),
    销售金额之和=('销售金额', 'sum'),
    平均单价=('单价', 'mean')
)

var8
```
