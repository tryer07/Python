```python
import pandas as pd

pd.set_option('display.min_rows',30) #这里默认是10行

#1.读取数据

df = pd.read_csv('data/sales.csv')

var1 = df

#2.数据清洗

#2.1缺失值处理

#2.1.1查找是否存在缺失值

var2 = df.isnull()

#2.1.2处理缺失值(填充或删除)

#删除缺失值

var3 = df.dropna()                     #删除缺失值所在行
var4 = df.dropna(axis = 1)             #删除缺失值所在列  ← 修正：drop → dropna

#填充缺失值

var5 = df.fillna('上海')                #填充缺失值(填充指定的缺失值)
var6 = df.ffill()                      #填充缺失值(使用上一行的数据填充)
var7 = df.bfill()                      #填充缺失值(使用下一行的数据填充)

# print(var3)
# print(var4)
# print(var5)
# print(var6)
# print(var7)
```

```python
#2.2重复值的处理

#2.2.1查看重复值

var8 = df.duplicated() #所有列的数据都重复才会判断为重复值
var9 = df.duplicated(subset=['订单号']) #指定列的数据重复才会判断为重复值，第二次以及以后出现的会返回True

# print(var8)
# print(var9)

#2.2.2删除重复值

var11 = df.drop_duplicates(subset=['订单号'], keep='first') #删除重复值，默认保留第一次出现的，keep='last'保留最后一次出现的，keep=False不保留任何重复值

# print(var11)
```

```python
#2.3异常值处理

#2.3.1查看异常值

var12 = df[df['单价'] < 0]

# print(var12)

#2.3.2删除异常值

var13 = df.drop(df[df['单价'] < 0].index)

# print(var13)

#2.3.3修复异常值

df['单价'] = df['单价'].abs() #求绝对值

# print(df)
```

```python
#2.4数据格式处理

#方法一

df['订单日期'] = df['订单日期'].replace('2025/06/02','2025-06-02') #仅支持修改单一值，并不推荐

# df

#方法二(统一修改，推荐使用)

df['订单日期'] = df['订单日期'].str.replace('/','-')

df
```
