```python
import pandas as pd

#创建Data Frame的多种方式

df1 = pd.DataFrame(
    [
        {"name": "Alice", "age": 25},
        {"name": "Bob", "age": 30},
        {"name": "Charlie", "age": 35}
    ]
)
df1
```

```python

df2 = pd.DataFrame(
    {
        "name": ["Alice", "Bob", "Charlie"],
        "age": [25, 30, 35]
    }
)
df2
```

```python

df3 = pd.DataFrame(
    [
        ['Alice', 25],
        ['Bob', 30],
        ['Charlie', 35]
    ],
    columns=["name", "age"]
)
df3
```

```python

df4 = pd.DataFrame(
    [
        ['Alice', 25],
        ['Bob', 30],
        ['Charlie', 35]
    ],
    columns=["name", "age"],
    index = ['a', 'b', 'c'],
)

df4
```
