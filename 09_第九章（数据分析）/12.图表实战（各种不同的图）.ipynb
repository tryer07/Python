```python
import matplotlib.pyplot as plt

# 设置中文字体，防止标题和坐标轴中文显示为方块
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 数据：不同国家的世界石油储备（单位：亿桶，示例数据）
countries = ['委内瑞拉', '沙特', '加拿大', '伊朗', '伊拉克', '科威特', '阿联酋', '俄罗斯']
values = [303, 267, 174, 156, 145, 105, 100, 80]

# 1. 创建一个一行两列的子图画布
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 6), dpi=100)

# 2. 获取第一个子图对象：axes[0]

# 3. 在第一个子图中绘制"世界石油储备"柱状图
axes[0].bar(countries, values)

# 4. 美化图表
# 标题（含字体大小）
axes[0].set_title('世界石油储备', fontsize=16)
# X 轴标签
axes[0].set_xlabel('国家', fontsize=12)
# Y 轴标签
axes[0].set_ylabel('石油储备（亿桶）', fontsize=12)
# 柱子颜色与宽度（在 bar 中设置）
axes[0].bar(countries, values, color='skyblue', width=0.6)

# 网格线：虚线 + 透明度
axes[0].grid(True, linestyle='--', alpha=0.5)

plt.show()
```
