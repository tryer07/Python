```python
import matplotlib.pyplot as plt

# 设置中文字体，防止标题和图例中的中文显示为方块
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 数据：世界石油储备排名靠前的主要国家（单位：亿桶，示例数据）
countries = ['委内瑞拉', '沙特', '加拿大', '伊朗', '伊拉克', '科威特', '阿联酋', '俄罗斯']
values = [303, 267, 174, 156, 145, 105, 100, 80]

# 数据：上述国家的人口（单位：百万人，示例数据）
population_values = [28, 36, 39, 88, 44, 4, 10, 144]

# 1. 创建一个一行两列的子图画布
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 6), dpi=100)

# ---------- 第一个子图：世界石油储备柱状图 ----------

# 2. 获取第一个子图对象：axes[0]，一次调用完成绘制、颜色与宽度设置
axes[0].bar(countries, values, color='skyblue', width=0.6)

# 3. 美化图表：标题、坐标轴标签、网格线
axes[0].set_title('世界石油储备排名靠前的主要国家', fontsize=16)
axes[0].set_xlabel('国家', fontsize=12)
axes[0].set_ylabel('石油储备（亿桶）', fontsize=12)
axes[0].grid(True, linestyle='--', alpha=0.5)

# ---------- 第二个子图：主要石油储备国人口比例饼状图 ----------

# 1. 获取画布中的第二个子图（索引为 1）
ax2 = axes[1]

# 2 & 3. 绘制饼状图，labels 显示国家名，autopct 显示百分比（%% 转义输出百分号）
#        返回值 pie 中包含扇形对象 wedges，供图例使用
pie = ax2.pie(population_values, labels=countries, autopct="%1.1f%%")

# 4. 设置饼状图标题（注明百分比口径：仅在所示石油储备大国之间比较，避免误导为世界人口比例）
ax2.set_title("主要石油储备国的人口比例\n（所示国家之间的占比，并非占世界总人口）")

# 5. 添加并优化图例：位置在下方居中、5 列、偏移锚点
ax2.legend(pie[0], countries, loc="lower center", ncol=5, bbox_to_anchor=(0.5, -0.1))

# 6. 保存图表为图片（必须在 plt.show() 之前，否则可能保存空白图）
plt.savefig("data/01.png")

plt.show()
```
