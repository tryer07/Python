```python
import matplotlib.pyplot as plt

import matplotlib.pyplot as plt

# 设置中文字体，防止标题和图例中的中文显示为方块
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei']
plt.rcParams['axes.unicode_minus'] = False

# x 轴：一天 1~24 小时
hours = list(range(1, 25))

# 2026年10月19日 逐小时温度（单位：℃）
# 参考当日真实预报：上海 阵雨转阴 21~24℃，福州 多云 25~33℃
shanghai_temps = [22, 22, 21, 21, 21, 21, 21, 22, 22, 23, 23, 24, 24, 24, 24, 23, 23, 22, 22, 22, 22, 22, 22, 22]
fuzhou_temps = [27, 26, 26, 25, 25, 25, 26, 28, 29, 30, 31, 32, 33, 33, 32, 31, 30, 29, 28, 28, 27, 27, 26, 26]

# 创建画布
plt.figure(figsize=(12, 6))

# 折线图：上海用蓝色，福州用绿色，label 用于图例显示
plt.plot(hours, shanghai_temps, color='blue', marker='o', label='上海市')
plt.plot(hours, fuzhou_temps, color='green', marker='o', label='福州市')

# 标题与坐标轴标签
plt.title('2026年10月19日 上海市与福州市逐小时温度对比')
plt.xlabel('时间（小时）')
plt.ylabel('温度（℃）')

# x 轴刻度显示 1~24 的每一个小时
plt.xticks(hours)

# 显示图例
plt.legend()

# 背景网格线
plt.grid(True)

# 展示图表
plt.show()
```
